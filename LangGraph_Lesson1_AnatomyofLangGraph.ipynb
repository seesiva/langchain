🧠 Lesson 1: The Anatomy of LangGraph
At its core, LangGraph turns standard AI scripts into a structured flowchart made of three distinct pieces

📝 State (The Shared Notebook): A persistent memory bucket (usually a TypedDict) that holds your data.

🤖 Nodes (The Workers): Standard Python functions that read the State, perform a specific task, and write their updates back to the notebook.

➡️ Edges (The Wires): The pathways that dictate the strict order of execution from START to END.

# 🧪 First LangGraph Agent

In this lesson, we will build the smallest possible LangGraph agent.

This graph has:

- **State**: shared memory
- **Node**: worker function
- **Edges**: execution flow

The flow is:

START → chatbot → END

The chatbot node will read the current State and add one AI message to the conversation.    Worker --> END((END)):::startEnd

Step 1: Install LangGraph

````python
!pip install -q -U langgraph

import os
from google.colab import userdata

os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
````

**Step 2: Imports**


````python
from typing import Annotated
from typing_extensions import TypedDict

from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
````

**Step 3: Define State**

State is the shared memory of the graph.

Here, the State contains one field: `messages`.

We use `add_messages` so that new messages are appended to the existing conversation instead of replacing it.

````python
class State(TypedDict):
    messages: Annotated[list, add_messages]
````

**Step 4: Define the Node**

A node is a normal Python function.

It receives the current State, performs one task, and returns updates to the State.

Here, the chatbot node adds a simple AI response.

````python
def simple_chatbot(state: State):
    return {
        "messages": [
            ("ai", "Hello! I am a basic LangGraph agent.")
        ]
    }
````

**Step 5: Build the Graph**

Now we create a LangGraph workflow using `StateGraph`.

Then we add the chatbot node to the graph.

````python
workflow = StateGraph(State)

workflow.add_node("chatbot", simple_chatbot)
````

**Step 6: Add Edges**

Edges define the execution flow.

In this basic example, the graph starts, runs the chatbot node, and then ends.

START → chatbot → END

````python
workflow.add_edge(START, "chatbot")
workflow.add_edge("chatbot", END)
````

**Step 7: Compile and Run**

Before running the graph, we compile it.

Then we invoke the graph with an initial human message.

````python
app = workflow.compile()

result = app.invoke({
    "messages": [
        ("human", "Hi there!")
    ]
})
````

**Step 8: Inspect the Final State**

The graph returns the final State.

Because we used `add_messages`, the final State contains both:

- the original human message
- the AI message added by the chatbot node

````python
for message in result["messages"]:
    print(f"{message.type.upper()}: {message.content}")
````
