*Before we learn how to use LangChain, we must understand why we need it.
Out of the box, Large Language Models (LLMs) are incredibly intelligent, but they suffer from two major limitations:*

1. **Frozen Knowledge**: They do not have live internet access. They only know the
world as it existed up until their training cutoff date.
2. **Mathematical Constraints**: LLMs generate text by predicting the next most likely word (autoregressive generation). They cannot natively compute exact, complex math.

In this exercise, we will expose these limitations by asking a raw, isolated LLM a question that requires both real-time knowledge and exact math: **"What is the square root of the current US president's age multiplied by 132?"**

```python
# 1. Install the base OpenAI library (No LangChain here!)
!pip install -qU openai
```

```python
from openai import OpenAI
from google.colab import userdata
api_key=userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=api_key,) # It automatically looks for the OPENAI_API_KEY env var
```

```python
# 4. Define our complex question
question = "What is the square root of the current US president's age multiplied by 132?"

print(f"Sending the question: '{question}' to the isolated LLM...\n")
```

```python
# 5. Ask the model using the standard API formatting
response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[{"role": "user", "content": question}],
    temperature=0 # Notice that we are setting the temperature to 0.
    #This makes the model's output as deterministic and analytical as possible, removing creative guessing.
)
```

```python
# 6. Display the output and explain the failure
# Use the correct attribute path (choices[0]) and wrap it in an f-string
print(f"--- Pure OpenAI API Response ---\n{response.choices[0].message.content}")

# Defining the limitations cleanly
reason_1 = "outdated training data to identify the president"
reason_2 = "fail the square root multiplication because it is generating text, not running a precise calculator. Actual value would be 102.11757928975794126513876456956"

print("\n--- Why this fails ---")
print(f"1. The model might use {reason_1}.")
print(f"2. The model will likely {reason_2}.")
```
